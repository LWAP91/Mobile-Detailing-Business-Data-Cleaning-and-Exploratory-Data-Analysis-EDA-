After visualizing the dataset , some data types are incorrect and there are also inconsistent data in many columns . before before starting the analysis, we must clean the dataset for better result.

Cleanning the date column by using pd.to_datetime method to parse it.

```python
# Making a copy of the Date column for safety inspection
Date_column = df_BNdetailing['Date'].copy()

# Parsing the date column copied with multiple formats
Date_parse = pd.to_datetime(Date_column, format='mixed', dayfirst=True)
```

```python
#  Checking for remaining NaT values if existed
Date_parse.isna().sum()

```

```python
# Updating the 'Date' column in the main DataFrame
df_BNdetailing['Date'] = Date_parse

# Verifying the update by checking the DataFrame info
df_BNdetailing.info()
```

Cleanning the final price column by convering it to a numeric data using pd.to_numeric function.

```python
# Copying the Final_Price column before the convertion
final_price_column = df_BNdetailing['Final_Price'].copy()

# Removing the '$' sign and converting  to numeric column
final_price_numeric = pd.to_numeric(final_price_column.astype(str).str.replace('$', '', regex=False), errors='coerce')

# Checking for NaN values if existed and the data types
final_price_numeric.isna().sum()
final_price_numeric.info()

# Updating the final price column to the main dataframe
df_BNdetailing['Final_Price'] = final_price_numeric
```

Cleanning the final price column by converting it to a numeric data using pd.to_numeric function

```python
# Making a copy of the column for safety inspection
total_amount_column = df_BNdetailing['Total_Amount'].copy()

# Converting the column and removing the $ sign
total_amount_numeric = pd.to_numeric(total_amount_column.astype(str).str.replace('$', '', regex=False), errors='coerce')

# Checking for NaN values if existed and the data type
total_amount_numeric.isna().sum()
total_amount_numeric.info()

# Updating the total_amount column
df_BNdetailing['Total_Amount'] = total_amount_numeric

df_BNdetailing.info()
```

Converting the data type of the tip column

```python
# Converting 'Tip' column from int64 to float64
df_BNdetailing['Tip'] = df_BNdetailing['Tip'].astype(float)

# Verify the data type conversion
df_BNdetailing['Tip'].info()

# Fixing the negative values
df_BNdetailing['Tip'] = df_BNdetailing['Tip'].abs()
```

Cleanning the customer name

```python
# Making a copy of the customer name column
customer_name_column = df_BNdetailing['Customer_Name'].copy()

# Removing any leading/trailing white spaces from the customer name
customer_name_column = customer_name_column.str.strip()

# Normalizing internal spaces (replace multiple spaces with a single space)
customer_name_column = customer_name_column.str.replace(r'\s+', ' ', regex=True)

# Converting the names in title format
customer_name_column = customer_name_column.str.title()

# Cheking for non string values in the column name
non_string_names = customer_name_column[customer_name_column.apply(lambda x: not isinstance(x, str))]
if non_string_names.empty:
    print("No non-string values found in the 'Customer_Name' column.")
else:
    print(f"Non-string value found: {non_string_names}")


# Updating the customer_name column to the main data frame
df_BNdetailing['Customer_Name'] = customer_name_column

# Visualizing the customer name column after the cleaning
df_BNdetailing['Customer_Name']
```

Cleanning the channel column

```python
# Making a copy of the channel column
channel_column = df_BNdetailing['Channel'].copy()

# Normalizing all text as string , removing leading and trailling space, and format all text in lower case.
channel_column = channel_column.astype(str).str.strip().str.lower()

# Mapping the messy values to clean categories
channel_mapping = {'instagram': 'Instagram','insta': 'Instagram',
                  'facebook': 'Facebook','fb': 'Facebook',
                  'youtube': 'Youtube', 'yotube':'Youtube',
                  'tiktok':'TikTok','tik tok': 'TikTok',
                  'referral': 'Referral', 'referred':'Referral',
                  'word of mouth':'Referral',
                  'social': 'Other'}
channel_column = channel_column.map(channel_mapping)

# Filling NaN values (unmapped channels) with 'Unknown'
channel_column = channel_column.fillna('Unknown')

# Updating the 'Channel' column in the main DataFrame with the cleaned values
df_BNdetailing['Channel'] = channel_column

# Displaying unique unmapped values (NaNs)
unmapped_channels = df_BNdetailing['Channel'][df_BNdetailing['Channel'].isna()].unique()
if len(unmapped_channels) > 0:
    print("Unmapped channels:\n", unmapped_channels)
else:
    print("All channels are mapped.")

# Displaying the unique values in the updated channel column
df_BNdetailing['Channel'].unique()
```

Cleaning the payment method colummn

```python
# Making a copy of the payment method column and visualizing the unique values
payment_method_column = df_BNdetailing['Payment_Method'].copy()

# Visualizing the unique values of the payment method
payment_method_column.unique()

# Normalizing the values as string and lowercase format
payment_method_column = payment_method_column.astype(str).str.strip().str.lower()

# Mapping the messy values to clean category
payment_method_mapping = {'cash':'Cash', 'venmo':'Venmo','cash app':'Cash App','card':'Card','zelle':'Zelle'}
payment_method_column = payment_method_column.map(payment_method_mapping)

# Displaying the unique unmapped values
unmapped_payment_method = df_BNdetailing['Payment_Method'][payment_method_column.isna()].unique()
if len(unmapped_payment_method) > 0:
    print("Unmapped payment methods:\n", unmapped_payment_method)
else:
    print ('all values are mapped')

# Updating the payment_method column to the main dataframe
df_BNdetailing['Payment_Method'] = payment_method_column
```

Fixing duplicated rows

```python
# Finding duplicate rows and displaying the amount
duplicate_rows = df_BNdetailing.duplicated().sum()
print(f"Number of duplicate rows: {duplicate_rows}")

# Removing the duplicated rows
df_BNdetailing = df_BNdetailing.drop_duplicates()
```

Handling missing values

```python
# Checking the missing values in the main dataframe
df_BNdetailing.isna().sum()

# Making a copy of the Rating column
rating_column = df_BNdetailing['Rating'].copy()

# Handling missing values in "Rating" column by filling NaN with "No Feedback"
rating_column.fillna('No Feedback', inplace=True)

# Updating the rating column to the main dataframe
df_BNdetailing['Rating'] = rating_column

# Making a copy of the Discount_Code column
discount_code_column = df_BNdetailing['Discount_Code'].copy()

# Handling missing values in "Discount_Code" column by filling NaN with 'No Discount'
discount_code_column.fillna('No Discount', inplace=True)

# Updatting the Discount_Code column to the main datafframe
df_BNdetailing['Discount_Code'] = discount_code_column


```

Creating the clean and final dataset

```python
# Creating the clean and final dataset
df_BNdetailing.to_csv('BNdetailing_2025_clean.csv', index=False)
```
